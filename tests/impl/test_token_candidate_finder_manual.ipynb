```python
from func_sketch._impl.token_candidate_finder import TokenCandidateFinder
from func_sketch._cpp import ExpressionParser

math_function_list = ExpressionParser().math_function_list()
constant_list = ExpressionParser().constant_list()
parameter_names = ["x"]
limit = 10
finder = TokenCandidateFinder(
    math_function_list=math_function_list,
    constant_list=constant_list,
    parameter_names=parameter_names,
    limit=limit,
)


def test_finder(query: str) -> None:
    """Test with a query string."""
    results = finder.find(query=query)
    str_list = [result.name for result in results]
    print(str_list)
```
